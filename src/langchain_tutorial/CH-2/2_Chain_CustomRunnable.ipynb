```python
import os
from dotenv import load_dotenv
load_dotenv()

if os.environ.get("OPENAI_API_KEY"):
    print("bro it is working")


from langchain.chat_models import init_chat_model
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
```

### **Chain with Custom Runnable**

```python
template1 = ChatPromptTemplate.from_messages([
    ("system", "you are a helful assistant"),
    ("human", "{input}")
])
```

```python
llm_model = init_chat_model("gpt-5-mini")
```

```python
str_parse = StrOutputParser()
```

```python
from langchain_core.runnables import RunnableLambda

def random_function(text:str):
    return {"text" : text}

runnable_random_function = RunnableLambda(random_function)
print(runnable_random_function)
```

```python
template2 = ChatPromptTemplate.from_messages([
    ("system", "you are a social media post generator "),
    ("human","Create the small post for the following text for linkIn : {text}")
])
```

```python
llm_model = init_chat_model("gpt-5-mini")
```

```python
str_parse = StrOutputParser()
```

```python
chain = template1 | llm_model | str_parse | runnable_random_function | template2 | llm_model | str_parse
chain.invoke("what is the capital of Karnataka?")
```
