```python
import os
from dotenv import load_dotenv
load_dotenv()

if os.environ.get("OPENAI_API_KEY"):
    print("bro it is working")


from langchain.chat_models import init_chat_model
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
```

```python
template = ChatPromptTemplate.from_messages([
    ("system", "you are a helful assistant"),
    ("human", "{input}")
])
```

```python
llm_model = init_chat_model("gpt-5-mini")
```

```python
str_parse = StrOutputParser()
```

### **Manual invocation**

```python
ready_template = template.invoke({"input" : "what is the capital of Karnataka"}) 
print(ready_template)
llm_model.invoke(ready_template)
```

### **Chain Invocation**

```python
chain = template | llm_model | str_parse
chain.invoke({"input" : "what is the capital of Karntaka"})
```

```python
from langchain_core.runnables import RunnableSequence
chain1 = RunnableSequence(template, llm_model, str_parse)

chain1.invoke("what is the capital of Karntaka") 
```
