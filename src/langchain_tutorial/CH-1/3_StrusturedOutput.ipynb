```python
import os
from dotenv import load_dotenv
load_dotenv()

if os.environ.get("OPENAI_API_KEY"):
    print("bro it is working")

from langchain.chat_models import init_chat_model

llm_model = init_chat_model("gpt-5-mini")
```

```python
llm_model.invoke("Tell me a joke. Generate the output in kay value pair, key should be fact and value should be punchline").content
```

```python
from pydantic import BaseModel, Field

class llm_schema(BaseModel):
    setup: str = Field(description="The set up for the joke")
    punchline: str = Field(description="The punchline for the joke")
```

```python
obj = llm_schema(**{"setup" : "some setup", "punchline" : "some punchline"})
obj
```

```python
llm_structured_ouput = llm_model.with_structured_output(llm_schema)
llm_structured_ouput.invoke("Tell me a joke")
```
