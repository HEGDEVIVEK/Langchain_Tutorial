### **Messages**

```python
from langchain.messages import HumanMessage, AIMessage, SystemMessage
from langchain.chat_models import init_chat_model

llm_model = init_chat_model("gpt-5-mini")

messages = [
    SystemMessage(content="You are a helpful assistant."),
    HumanMessage(content="Explain what LangChain messages are in one sentence."),
]

response = llm_model.invoke(messages)
print(response.content)
```

### **Prompts**

```python
from langchain_core.prompts import ChatPromptTemplate

template = ChatPromptTemplate.from_template("write a fun factor about {topic}")

ready_prompt = template.invoke({"topic" : "pen"})
print(ready_prompt)

llm_model.invoke(ready_prompt).content
```

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain.chat_models import init_chat_model

llm_model = init_chat_model("gpt-5-mini")

template = ChatPromptTemplate.from_messages(
    [
        ("system", "you are {tone} assistant"),
        ("human", "write a fun factor about {topic} and reurn the output in such a way that key is fact and value is fun fact")
    ]
)

system_tone = input("Enter a tone")
user_topic = input("Enter a topic")

ready_prompt = template.invoke({"tone":system_tone,"topic":user_topic})
print(ready_prompt)

llm_model.invoke(ready_prompt).content
```
